```python
import pandas as pd
import numpy as np
import random

df = pd.read_csv('spotify_songs.csv')
```

```python
def consultar_canciones(cantidad=10):
    if cantidad is None:
        cantidad = 10
    return df[["track_name", "track_artist", "playlist_genre", "track_popularity"]].sort_values(
        by="track_popularity",
        ascending=False
    ).head(cantidad)

consultar_canciones(2)
```

```python
def crear_playlist(nombre):
    playlist = {
        "nombre": nombre,
        "canciones": []
    }

    return playlist

crear_playlist("Gym")
```

```python
def agregar_cancion(playlist, id_cancion):
    res = df[df["track_id"] == id_cancion]

    if res.empty:
        print("Cancion no encontrada")
        return playlist

    cancion = res.iloc[0]

    playlist["canciones"].append({
        "track_id": cancion["track_id"],
        "track_name": cancion["track_name"],
        "track_artist": cancion["track_artist"]
    })
    
    return playlist

agregar_cancion(crear_playlist("Gym"), "7o2CTH4ctstm8TNelqjb51")
```

```python
def consultar_playlist(playlist):
    print(f"Playlist: {playlist['nombre']}")
    print(f"Total de canciones: {len(playlist['canciones'])}")
    
    for i, cancion in enumerate(playlist["canciones"], 1):
        print(f"{i}. {cancion['track_name']} - {cancion['track_artist']}")

consultar_playlist(agregar_cancion(crear_playlist("Gym"), "7o2CTH4ctstm8TNelqjb51"))
```

```python
def eliminar_cancion(playlist, id_cancion):
    for cancion in playlist["canciones"]:
        if cancion["track_id"] == id_cancion:
            playlist["canciones"].remove(cancion)

    return playlist


playlist =crear_playlist("Gym")

consultar_playlist(playlist)

agregar_cancion(playlist, "7o2CTH4ctstm8TNelqjb51")
agregar_cancion(playlist, "22TOCwH0z9Mcwoa1tfQbro")
agregar_cancion(playlist, "5wO14EzcJEdnaDRmI35PEP")

consultar_playlist(playlist)

eliminar_cancion(playlist, "7o2CTH4ctstm8TNelqjb51")

consultar_playlist(playlist)
```
